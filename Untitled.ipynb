```python
import numpy as np
from SOAP import SOAP, PHOENIX
from SOAP import fast_starspot as stspnumba
from SOAP.SOAP import without_units
from SOAP.classes import Spectrum

psi_test = np.array([0.0])

# Observational parameters 
λ_TESS=[6000, 10000]    # Wavelength range [Angstrom] 

# Observational parameters 
λ_CHEOPS=[4000, 8500]    # Wavelength range [Angstrom] 

# Star-like star, rigid rotation
radius=1                # Stellar radius [solar radii]
mass=1                  # Stellar mass [solar masses]
prot=24.47              # Stellar rotation period [days]
incl=45                 # Stellar inclination [degrees] -> 90 is equator-on, 0 is pole-on
logg= 4.4               # Stellar surface gravity [cgs]
feh= 0.00               # Stellar metallicity [dex]
Teff=5777               # Stellar effective temperature [K]
ldcn=[0.5, 0.16]        # Linear and quadratic limb-darkening coefficients (must agree with λ, see below)
law=1                   # Limb-darkening law (quadratic)
start_psi=0.0           # Starting phase of the star

# generate spectra 
spQ_CHEOPS = PHOENIX(wave_range=λ_CHEOPS, normalize=False, teff=Teff, logg=logg, Z=feh) # stellar spectrum
spQ_ar_CHEOPS = PHOENIX(wave_range=λ_CHEOPS, normalize=False, teff=Teff, logg=logg, Z=feh, contrast=0, ar=True) # AR spectrum

# convolve resolution down to more realistic instrumentation values 
Spectrum.convolve_to(spQ_CHEOPS, 50000, inplace=True)
Spectrum.convolve_to(spQ_ar_CHEOPS, 50000, inplace=True)

sim_test = SOAP.Simulation(
    pixel=spQ_CHEOPS,
    pixel_ar=spQ_ar_CHEOPS,
    wlll=6250.0,
    grid=50,
    active_regions=[],
    resample_spectra=10,
)

sim_test.active_regions = [
    SOAP.ActiveRegion(
        check=True,
        lon=0.0,       # disk centre for an equator-on star
        lat=0.0,
        size=0.084,
        tempdiff=600,
        active_region_type=0,
    )
]

sim_test.star.set(
    prot=24.47,
    incl=90,           # equator-on: spot at lon=0 is fully visible
    coeffs=ldcn,
    law=1,
    start_psi=0.0,
    radius=1.0,
    mass=1.0,
    teff=5777,
)

sim_test.planet.Rp = 0.0
sim_test.planet.P = 1.0e6
sim_test.planet.t0 = -0.75e6
sim_test.planet.e = 0.0
sim_test.planet.w = 0.0
sim_test.planet.ip = 90.0
sim_test.planet.a = 0.05

sim_test.ring = SOAP.Ring(fi=1, fe=1, ir=90, theta=0)

ar_out_test = stspnumba.active_region_contributions(
    psi_test,
    without_units(sim_test.star),
    sim_test.active_regions,
    sim_test.pixel.to_numba(),
    sim_test.pixel_ar.to_numba(),
    sim_test.grid,
    sim_test.nrho,
    sim_test.wlll,
    without_units(sim_test.planet),
    without_units(sim_test.ring),
    ccf=False,
    skip_rv=False,
)

print("broadband flux:", ar_out_test[0])
print("total spectrum nonzero:", np.any(np.asarray(ar_out_test[3]) != 0.0))
```
