# Importing Libraries

```python
# import libraries
import numpy as np
import SOAP
from SOAP import PHOENIX
from IPython.display import HTML
from SOAP import fast_starspot as stspnumba
```

# Set Stellar Parameters

```python
# set parameters for Sun-like star with rigid rotation 
radius=1                # Stellar radius [solar radii]
mass=1                  # Stellar mass [solar masses]
prot=24.47              # Stellar rotation period [days]
incl=45                 # Stellar inclination [degrees] -> 90 is equator-on, 0 is pole-on
logg= 4.4               # Stellar surface gravity [cgs]
feh= 0.00               # Stellar metallicity [dex]
Teff=5777               # Stellar effective temperature [K]
ldcn=[0.5, 0.16]        # Linear and quadratic limb-darkening coefficients (must agree with λ, see below)
law=1                   # Limb-darkening law (quadratic)
start_psi=0.0           # Starting phase of the star
```

# Set Observational Parameters

## CHEOPS

```python
# Observational parameters 
λ_CHEOPS=[3300, 11000]    # Wavelength range [Angstrom] 
```

## TESS

```python
# Observational parameters 
λ_TESS=[6000, 10000]    # Wavelength range [Angstrom] 
```

# Perform CCF Simulation and Visualise Results

## Create Function

```python
# create function to run SOAP CCF simulation for different bandpasses
def run_sim_CCF(λ): 
    
    # create simulation variable with 600 grid points 
    # wavelength for temperature contrast via Planck's law is median of above list
    sim = SOAP.Simulation(wlll=(λ[1]+λ[0])/2, grid=600)

    # add active region at equator
    sim.active_regions=[SOAP.ActiveRegion(check=True, lon=180, lat=0, size=0.084, temp_diff=600, active_region_type=0)]

    # Set the properties of the star
    sim.star.set(prot=prot,incl=incl,coeffs=ldcn,law=law,start_psi=start_psi,radius=radius,mass=mass,teff=Teff)

    # remove planet
    sim.planet.Rp=0

    # Simulate over 3 stellar phases
    psi = np.linspace(0,3,6)

    # Calculate the RV and photometric time series
    out=sim.calculate_signal(psi, renormalize_rv=False)
    
    return sim, out
```

## CHEOPS

```python
# run simulation for CHEOPS
sim_CCF_CHEOPS, out_CCF_CHEOPS = run_sim_CCF(λ_CHEOPS)

# Create an animation of the flux variation
ani = sim_CCF_CHEOPS.visualize_animation(
    output=out_CCF_CHEOPS,
    plot_type="flux"
)
HTML(ani.to_jshtml())
```

```python
print(stspnumba.active_region_contributions(np.linspace(0,3,6), 
                                           sim_CCF_CHEOPS.star, 
                                           sim_CCF_CHEOPS.active_regions, 
                                           sim_CCF_CHEOPS.pixel, 
                                           sim_CCF_CHEOPS.pixel_ar, 
                                           sim_CCF_CHEOPS.grid, 
                                           sim_CCF_CHEOPS.nrho, 
                                           sim_CCF_CHEOPS.wlll, 
                                           sim_CCF_CHEOPS.planet, 
                                           ring=sim_CCF_CHEOPS.ring, 
                                           skip_rv=False))
```

## TESS

```python
# run simlation for TESS
sim_CCF_TESS, out_CCF_TESS = run_sim_CCF(λ_TESS)

# Create an animation of the flux variation
ani = sim_CCF_TESS.visualize_animation(
    output=out_CCF_TESS,
    plot_type="flux"
)
HTML(ani.to_jshtml())
```
